```python
import numpy as np

nums = input("Enter the elements: ").split()

numbers = []
for n in nums:
    numbers.append(int(n))

def sep_even_odd():
    numbers_np = np.array(numbers)
    even = numbers_np[numbers_np % 2 == 0]
    odd = numbers_np[numbers_np % 2 != 0]
    return even, odd

even_numbers, odd_numbers = sep_even_odd()

print("Even numbers:", even_numbers)
print("Odd numbers:", odd_numbers)
```
